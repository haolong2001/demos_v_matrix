## calculate immigration rates

### age-specific

```python
immigration_rates = []
for i in range(8):
    # Get dimensions
    rows, cols = migration_matrices[i].shape
    immigration_rate = np.zeros((rows, cols))
    
    # Convert migration matrix to numpy array for faster operations
    migration_array = migration_matrices[i].to_numpy()
    # Use numpy where to handle division by zero
    immigration_rate = np.where(scaled_popu[i][:,1:] != 0, 
                              migration_array / scaled_popu[i][:,1:],
                              0)
                
    immigration_rates.append(immigration_rate)

# Print first immigration rate matrix
print("Immigration rate matrix for first ethnic group:")
print(immigration_rates[0])
```

### across all the ages

```python
total_immigration_rates = []
for i in range(8):
    # Get dimensions
    rows, cols = migration_matrices[i].shape
    
    # Calculate column sums
    migration_colsum = np.sum(migration_matrices[i].to_numpy(), axis=0)
    population_colsum = np.sum(scaled_popu[i][:,1:], axis=0)
    
    # Calculate total immigration rate using numpy where to handle division by zero
    total_immigration_rate = np.where(population_colsum != 0,
                                    migration_colsum / population_colsum,
                                    0)
    
    total_immigration_rates.append(total_immigration_rate)

# Print total immigration rates for all ethnic groups
for i in range(8):
    print(f"Total immigration rate for ethnic group {i}:")
    print(total_immigration_rates[i])
```

# fit a uniform distribution to immigration rates

### we used rates from 2010 - 2019 (-14 to -4)

```python
immigration_rates[0][1,:]
```

```python
immigration_rates[0][1,-14:-4]
```

```python
# Initialize arrays to store uniform distribution parameters for all ethnic groups
uniform_params = np.zeros((8, 50, 2))  # Shape: (8 ethnic groups, 50 ages, 2 params)

# For each ethnic group
for i in range(8):
    # Get the immigration rates data for this ethnic group
    # -4 is the year 2019, pre-covid 
    imm_rates = immigration_rates[i][1:51,-14:-4]
    
    # For each row (age), fit uniform distribution to non-zero values
    for j, row in enumerate(imm_rates):
        non_zero_vals = row[row != 0]
        if len(non_zero_vals) > 0:
            # Get min and max for uniform distribution
            min_val = np.min(non_zero_vals)
            max_val = np.max(non_zero_vals)
            uniform_params[i,j] = [min_val, max_val]

print("Uniform distribution parameters shape:", uniform_params.shape)
print("Parameters for first ethnic group:")
print(uniform_params[i])

# Example of generating samples using the parameters
def generate_samples(params, n_samples=20):
    samples = np.zeros((8, 50, n_samples))
    for i in range(8):
        for j in range(50):
            min_val, max_val = params[i,j]
            if min_val != 0 or max_val != 0:
                samples[i,j] = np.random.uniform(min_val, max_val, n_samples)
    return samples

# Generate some sample data
sample_data = generate_samples(uniform_params)
print("\nGenerated samples shape:", sample_data.shape)
```

```python
import pandas as pd

# Create a comprehensive dataframe with uniform distribution parameters
# for all ethnic groups and ages

# Create lists to store data for DataFrame
data = []

# Iterate through all ethnic groups and ages
for ethnic_group in range(8):
    for age in range(50):
        data.append({
            'ethnic_group': ethnic_group,
            'age': age + 1,  # Ages 1-50
            'min_param': uniform_params[ethnic_group, age, 0],
            'max_param': uniform_params[ethnic_group, age, 1],
           
        })

# Create DataFrame
immigration_params_df = pd.DataFrame(data)

# Display basic information about the dataframe
print("Immigration Parameters DataFrame:")
print(f"Shape: {immigration_params_df.shape}")
print("\nFirst 10 rows:")
print(immigration_params_df.head(10))

# print("\nSample statistics:")
# print(immigration_params_df.describe())

# # Show parameters for a specific ethnic group
# print(f"\nParameters for ethnic group 0:")
# ethnic_0_df = immigration_params_df[immigration_params_df['ethnic_group'] == 0]
# print(ethnic_0_df.head(10))
```

# future immigration calculation

```python
# Initialize 8x86x27 numpy matrix for future immigration (8 groups, 86 age groups, 27 years from 2024-2050)
future_popu = np.zeros((8, 86, 28))  # from 2023 to 2050
future_immigration = np.zeros((8, 50, 27))
```

```python
# Convert immigration_params (400x5) to 8x50x2 numpy array
# Order: chinese male, chinese female, malay male, malay female, indian male, indian female, other male, other female
uniform_params = np.zeros((8, 50, 2))

races = ['chinese', 'malay', 'indian', 'other']
genders = ['male', 'female']

for race_idx, race in enumerate(races):
    for gender_idx, gender in enumerate(genders):
        group_idx = race_idx * 2 + gender_idx
        mask = (immigration_params['race'] == race) & (immigration_params['gender'] == gender)
        filtered_data = immigration_params[mask]
        
        # Extract lower and upper bounds for ages 1-50
        uniform_params[group_idx, :, 0] = filtered_data['lower'].values  # lower bounds
        uniform_params[group_idx, :, 1] = filtered_data['upper'].values  # upper bounds

uniform_params
```

```python
ages = np.arange(0, 85)
fertility_ages = range(15, 50)
for i in [1,0,3,2,5,4,7,6]: # deal with female first, then male 
    # get 2023 population
    popu_2023 = scaled_popu[i][:,-1] # last column
    future_popu[i,:,0] = popu_2023
    ## print(np.sum(immigration_rates[i], axis=0))

    for j in range(1,28): # 2024 to 2050

        immi_rate = np.array([
            np.random.uniform(low, high) 
            for low, high in uniform_params[i]
        ])

        
        # for the other ages (2, 18, 27)
        
        is_female = i % 2  # 1 is female, 0 is male
        year_index = j-1 # index 0 is year 2024
        # this is for mortality rate of 1 - 84
        mortaility_rates = mortality_forecast_from_24[is_female, ages // 5, j-1] / 1000.
        
        # age 0 - 84 --> 1 -- （85）, then experience death this year
        future_popu[i,1:86,j]=  (future_popu[i,0:85,j-1] * (1- mortaility_rates)).astype(int)

        # for 85 and 85+, add those people are still alive 
        mortaility_rate = mortality_forecast_from_24[is_female, 85 // 5, j-1] / 1000.
        future_popu[i,85,j] = (future_popu[i,85,j-1] * (1-mortaility_rate)).astype(int) +  future_popu[i,85,j]

        # then calculate immigration people
        
        future_immigration[i,:,j-1] = (future_popu[i,1:51,j] * immi_rate).astype(int)

        print("i =", i, "is_female =", is_female, 
                "sum immigration =", np.sum(future_immigration[i,:,j-1]))
        # add immigration back 
        future_popu[i,1:51,j] += future_immigration[i,:,j-1]

        # deal wtih fertility
        if i % 2 == 1:
            year = 2023 + j
            rates = [map_fertility_rate(i // 2, year, age, fertility_matrix) for age in fertility_ages]

            births = future_popu[i,15:50,j] * rates
            boys = (births * 1/ (1+ 1.06) ).astype(int)
            girls = (births - boys).astype(int)
            future_popu[i,0,j] =  np.sum(girls)
            future_popu[i-1,0,j] = np.sum(boys)

            # use the gender ratio to determine boys and girls

        
```
