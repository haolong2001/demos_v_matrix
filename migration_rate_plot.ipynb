```python
import numpy as np
import matplotlib.pyplot as plt

# Migration rates data
migration_rates = [0.0000, 0.0080, 0.0069, 0.0062, 0.0055, 0.0059, 0.0066, 0.0060, 
                  0.0064, 0.0063, 0.0050, 0.0067, 0.0077, 0.0009, 0.0050, 0.0054, 
                  0.0055, 0.0065, 0.0067, 0.0109, 0.0041, 0.0023, 0.0049, 0.0041, 
                  0.0049, 0.0054, 0.0045, 0.0060, 0.0050, 0.0059, 0.0042, 0.0030, 
                  0.0139, 0.0119]

migration_rates = migration_rates[:31]
# Create years array (1990-2020)
years = np.arange(1990, 2021)

# Convert rates to per 1000
migration_rates = [rate * 1000 for rate in migration_rates]
# Create the plot
plt.figure(figsize=(15, 6))
plt.plot(years, migration_rates, marker='o', linewidth=2, markersize=6)

# Customize the plot
# plt.title('Singapore Migration Trend (1990-2020)', fontsize=20, pad=15, weight='bold')
plt.xlabel('Year', fontsize=22)
plt.ylabel('Migration per 1000 Population', fontsize=20)

# Set major and minor ticks for x-axis
plt.gca().xaxis.set_major_locator(plt.MultipleLocator(5))  # Major ticks every 5 years
plt.gca().xaxis.set_minor_locator(plt.MultipleLocator(1))  # Minor ticks every year
plt.gca().xaxis.set_major_formatter(plt.FormatStrFormatter('%d'))

# Set major and minor ticks for y-axis
plt.gca().yaxis.set_major_locator(plt.MultipleLocator(2))  # Major ticks every 2 units
plt.gca().yaxis.set_minor_locator(plt.MultipleLocator(0.5))  # Minor ticks every 0.5 units

# Customize tick appearance
plt.tick_params(axis='both', which='major', labelsize=20, width=2, length=8)
plt.tick_params(axis='both', which='minor', width=1, length=4)

# Rotate x-axis labels for better readability
plt.xticks(rotation=0)

# Add horizontal line at y=0
# plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Show the plot
plt.show()

# Print statistical summary
print("\nStatistical Summary:")
print("-" * 40)
print(f"Mean: {np.mean(migration_rates):.4f}")
print(f"Median: {np.median(migration_rates):.4f}")
print(f"Standard Deviation: {np.std(migration_rates):.4f}")
print(f"Maximum: {np.max(migration_rates):.4f}")
print(f"Minimum: {np.min(migration_rates):.4f}")
```
