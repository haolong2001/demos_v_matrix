```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error

# Initialize arrays for storing results
mortality_matrix_mat = np.zeros((2, 86, 30))  # For storing forecasts
rmse_scores = []  # For storing RMSE scores

for i in [0,1]:  # Loop for both groups
    # Split data into train and validation sets
    train_data = filtered_matrix[i, :, :20]  # First 20 years for training
    val_data = filtered_matrix[i, :, 20:]    # Last years for validation
    
    # Step 1: Fit Lee-Carter on training data
    log_mortality = np.log(train_data)
    a_x = np.mean(log_mortality, axis=1)  # Average over columns (time)
    log_mortality_centered = log_mortality - a_x[:, np.newaxis]  # Center the data

    # Singular Value Decomposition (SVD)
    U, S, Vt = np.linalg.svd(log_mortality_centered, full_matrices=False)
    b_x = U[:, 0]  # First singular vector
    k_t = S[0] * Vt[0]  # First singular value multiplied by the first row of Vt

    # Step 2: Forecast k_t using ARIMA
    model = ARIMA(k_t, order=(1, 1, 0))
    model_fit = model.fit()
    k_t_forecast = model_fit.forecast(steps=val_data.shape[1])  # Forecast for validation period

    # Step 3: Reconstruct mortality rates for validation period
    forecasted_mortality = np.exp(a_x[:, np.newaxis] + np.outer(b_x, k_t_forecast))
    
    # Step 4: Calculate RMSE between forecasted and actual values
    rmse = np.sqrt(mean_squared_error(val_data.flatten(), forecasted_mortality.flatten()))
    rmse_scores.append(rmse)
    
    # Step 5: Now generate full 10-year forecast for mortality_matrix_mat
    k_t_full_forecast = model_fit.forecast(steps=10)
    forecasted_mortality_full = np.exp(a_x[:, np.newaxis] + np.outer(b_x, k_t_full_forecast))
    
    # Store in mortality_matrix_mat (only 10 years forecast)
    mortality_matrix_mat[i, 0, :10] = forecasted_mortality_full[0, :10]
    mortality_matrix_mat[i, 85, :10] = forecasted_mortality_full[-1, :10]
    
    for age in range(10, 71, 10):
        mortality_matrix_mat[i, age, :10] = forecasted_mortality_full[age//4, :10]

print(f"RMSE for Male (i=0): {rmse_scores[0]:.6f}")
print(f"RMSE for Female (i=1): {rmse_scores[1]:.6f}")
print("\nMortality matrix shape:", mortality_matrix_mat.shape)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

# Assuming mortality_matrix_mat and filtered_matrix are already defined
# Ensure mortality_matrix_mat has the correct shape
mortality_matrix_mat = np.zeros((2, 86, 30))  # Ensure (2, 86, 30)

for i in [0,1]:  # Loop for both groups
    # Step 1: Extract mortality data for all 30 forecasted columns
    mortality_matrix = filtered_matrix[i, :, :30]  # Shape (n, 30)

    # Step 2: Fit the Lee-Carter Model
    log_mortality = np.log(mortality_matrix)  # Log transformation
    a_x = np.mean(log_mortality, axis=0)  # Average over rows (time)
    log_mortality_centered = log_mortality - a_x  # Center the data

    # Singular Value Decomposition (SVD)
    U, S, Vt = np.linalg.svd(log_mortality_centered, full_matrices=False)
    b_x = U[:, 0]  # First singular vector
    k_t = S[0] * Vt[0]  # First singular value multiplied by the first row of Vt

    # Step 3: Forecast k_t using ARIMA
    model = ARIMA(k_t, order=(1, 1, 0))  # ARIMA(1,1,0) model
    model_fit = model.fit()
    k_t_forecast = model_fit.forecast(steps=30)  # Forecast for 30 years, start from 2021 - 2050

    # Step 4: Reconstruct future mortality rates
    forecasted_mortality = np.exp(a_x + np.outer(b_x, k_t_forecast))  # Shape (n, 30)

    # Step 5: Restore data in mortality_matrix_mat
    mortality_matrix_mat[i, 0, :] = forecasted_mortality[0, :]  # Restore row 0
    mortality_matrix_mat[i, 85, :] = forecasted_mortality[-1, :]  # Restore row 85

    # Restore other rows by expanding each row to 4 rows
    for idx in range(1, 23):  # Exclude first (0) and last (85)
        start_idx = 1 + idx * 4 -4
        end_idx = min(start_idx + 4, 85)  # Ensure within bounds
        mortality_matrix_mat[i, start_idx:end_idx, :] = forecasted_mortality[idx, :]

print(mortality_matrix_mat.shape)  # Should be (2, 86, 30)
```
